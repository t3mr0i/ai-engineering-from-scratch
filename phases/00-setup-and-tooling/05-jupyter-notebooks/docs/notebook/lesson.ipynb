# Jupyter Notebooks

> Treat the kernel namespace and restart boundary as part of the experiment.

## Build It: local kernel contract

```python
from main import NotebookKernel

kernel = NotebookKernel()
for source in ["samples = [2, 4, 8]", "print(samples)", "sum(samples) / len(samples)"]:
    print(kernel.execute(source))
kernel.restart()
print(kernel.execute("samples").error_type)
```

The local `NotebookKernel` uses `ast`, `exec`, and a shared namespace. It captures stdout, displays the last expression, and returns structured errors. A real IPython/Jupyter interface adds magics and rich display, but those are not dependencies of this stdlib simulation.

## Use It: restart and run all

```python
results = kernel.run_all(["x = 3", "x *= 4", "x"])
print(results[-1].display)
print(kernel.execute("1 / 0").error_type)
```
