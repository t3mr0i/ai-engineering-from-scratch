This notebook exercises the local pipeline implementation and its stated contracts.


```python
from pathlib import Path
import sys
candidates = [Path.cwd() / 'code', Path.cwd().parent.parent / 'code', Path.cwd() / 'phases/02-ml-fundamentals/13-ml-pipelines/code']
code_dir = next(p for p in candidates if (p / 'pipeline.py').exists())
sys.path.insert(0, str(code_dir))
import pipeline
```

```python
data = pipeline.make_mixed_data(40, seed=2)
train, test = pipeline.train_test_split_dict(data, test_ratio=0.25, seed=2)
model = pipeline.FullPipeline(pipeline.DecisionTreeSimple(max_depth=2), ['age', 'income', 'score'], ['city', 'plan'])
model.fit(train)
predictions = model.predict(test)
assert predictions.shape == test['target'].shape
encoder = pipeline.OneHotEncoder(handle_unknown='error').fit([['red'], ['blue']])
try:
    encoder.transform([['green']])
except ValueError:
    pass
else:
    raise AssertionError('unknown category was silently accepted in error mode')
print({'train_rows': len(train['target']), 'test_rows': len(test['target']), 'prediction_shape': predictions.shape})
```
