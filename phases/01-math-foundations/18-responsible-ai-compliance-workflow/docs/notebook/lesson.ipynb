```python
from pathlib import Path
import sys
candidates = [Path.cwd() / '../../code', Path.cwd() / 'code', Path('phases/01-math-foundations/18-responsible-ai-compliance-workflow/code')]
for candidate in candidates:
    if (candidate / 'main.py').exists():
        sys.path.insert(0, str(candidate.resolve()))
        break
from main import Scenario, recommend, build_plan, demo_scenarios
```

```python
scenario = Scenario('HR screening assistant', 'Employee information feeds an automated decision and an explanation is required for affected staff.', ('sensitive data', 'automated decision', 'explanation required'), impact=4, uncertainty=4)
recommendation = recommend(scenario)
print('signals:', ['sensitive data', 'automated decision', 'explanation required'])
print('categories:', recommendation.categories)
print('score/priority:', recommendation.score, recommendation.priority)
print('controls:', recommendation.controls)
print('evidence:', recommendation.evidence)
assert recommendation.score == 18
assert 'appeal path' in recommendation.controls
```
