## Primitive transforms

Apply each matrix to the same point and keep the determinant beside the result.

```python
p = [1.0, 0.0]
r = rotation_2d(3.141592653589793 / 2)
s = scaling_2d(2, 0.5)
mat_vec_mul(r, p), mat_vec_mul(s, [1.0, 1.0]), det_2x2(r), det_2x2(s)
```

Rotation preserves orientation and area; the selected scaling also has determinant one because `2*0.5=1`.

```python
mat_vec_mul(mat_mul(s, r), p), mat_vec_mul(mat_mul(r, s), p)
```

The two compositions differ because the rightmost matrix acts first. For `[[2,1],[1,2]]`, the eigenvalues are `1` and `3`.

```python
sorted(eigenvalues_2x2([[2, 1], [1, 2]]))
```
