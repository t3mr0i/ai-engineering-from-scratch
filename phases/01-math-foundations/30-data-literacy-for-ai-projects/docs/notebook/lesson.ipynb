```python
from pathlib import Path
import sys
candidates = [Path.cwd() / '../../code', Path.cwd() / 'code', Path('phases/01-math-foundations/30-data-literacy-for-ai-projects/code')]
for candidate in candidates:
    if (candidate / 'main.py').exists():
        sys.path.insert(0, str(candidate.resolve()))
        break
from main import Scenario, recommend, build_plan, demo_scenarios, signal_matches
```

```python
scenario = Scenario('customer table', 'The extract is being assessed before a pilot.', (), impact=4, uncertainty=4, source_owner=None, freshness_days=91, quality_rate=0.80, sensitive_fields=('email',))
matches = signal_matches(scenario)
recommendation = recommend(scenario)
print('derived signals:', matches)
print('categories:', recommendation.categories)
print('score/priority:', recommendation.score, recommendation.priority)
print('controls:', recommendation.controls)
print('evidence:', recommendation.evidence)
assert matches == ['unclear source owner', 'stale data', 'quality issue', 'sensitive field']
assert recommendation.score == 20
```

```python
plan = build_plan(demo_scenarios())
print('handoff rows:')
for row in plan:
    print(row['scenario'], row['signals'], row['categories'], row['score'])
assert plan == sorted(plan, key=lambda row: (-row['score'], row['scenario']))
assert all({'signals', 'categories', 'controls', 'evidence'} <= row.keys() for row in plan)
```
