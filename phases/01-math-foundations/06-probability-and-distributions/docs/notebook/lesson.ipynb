## PMFs and PDFs

A PMF assigns mass to a discrete outcome; a PDF is a density whose interval integral is probability.

```python
bernoulli_pmf(0, 0.7), bernoulli_pmf(1, 0.7), poisson_pmf(0, 3)
```

The Bernoulli masses are `0.3` and `0.7`; the Poisson zero mass is `exp(-3)`, approximately `0.0498`.

```python
small = softmax([2.0, 1.0, 0.1])
large = softmax([1000.0, 1001.0])
small, sum(small), large, sum(large), cross_entropy_loss([2.0, 1.0, 0.1], 0)
```

Subtracting the maximum logit keeps both softmax calls finite. The target index for this Python helper is zero-based, matching the lesson's cross-entropy fixture.
