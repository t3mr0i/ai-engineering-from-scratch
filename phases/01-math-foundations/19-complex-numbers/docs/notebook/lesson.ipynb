```python
from pathlib import Path
import sys
candidates = [Path.cwd() / '../../code', Path.cwd() / 'code', Path('phases/01-math-foundations/19-complex-numbers/code')]
for candidate in candidates:
    if (candidate / 'complex_numbers.py').exists():
        sys.path.insert(0, str(candidate.resolve()))
        break
from complex_numbers import Complex, euler, dft, idft
```

```python
z = Complex(3, 4)
print('z:', z, 'magnitude:', z.magnitude())
print('z * conjugate:', z * z.conjugate())
rotated = euler(3.141592653589793 / 2)
print('quarter-turn:', rotated)
```

```python
signal = [Complex(1, 0), Complex(2, 0), Complex(1, 0), Complex(0, 0)]
spectrum = dft(signal)
recovered = idft(spectrum)
print('spectrum:', spectrum)
print('recovered:', recovered)
assert all(abs(a.real - b.real) < 1e-9 for a, b in zip(signal, recovered))
```
