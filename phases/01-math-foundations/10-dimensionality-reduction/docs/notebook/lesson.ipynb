# Dimensionality Reduction

The canonical NumPy implementation centers a local fixture, eigendecomposes its covariance, and reconstructs it.

```python
from pathlib import Path
import sys
candidates = (Path.cwd() / 'code', Path.cwd().parents[1] / 'code', Path.cwd() / 'phases/01-math-foundations/10-dimensionality-reduction/code')
code_dir = next(path for path in candidates if (path / 'dim_reduction.py').is_file())
sys.path.insert(0, str(code_dir))
from dim_reduction import PCA, kernel_pca, make_concentric_circles, make_synthetic_data, reconstruction_error
import numpy as np
```

## Build It: covariance PCA

For 60 rows and three features, two components produce a `(60, 2)` projection. The retained ratios are the selected covariance eigenvalues divided by total variance.

```python
X = make_synthetic_data(seed=7, n_samples=60)
pca = PCA(2)
Z = pca.fit_transform(X)
X_hat = pca.inverse_transform(Z)
X.shape, Z.shape, np.round(pca.explained_variance_ratio_, 4), round(reconstruction_error(X, X_hat), 6)
```

## Use It: nonlinear similarity

The concentric-circle fixture is not separated by one linear axis. Centered RBF kernel PCA returns one row per point and two nonlinear coordinates.

```python
circles, labels = make_concentric_circles(n_per_ring=20)
linear = PCA(1).fit_transform(circles)
rbf = kernel_pca(circles, n_components=2, gamma=0.5)
linear.shape, rbf.shape, labels.tolist()[:4], np.round(rbf.mean(axis=0), 8)
```
