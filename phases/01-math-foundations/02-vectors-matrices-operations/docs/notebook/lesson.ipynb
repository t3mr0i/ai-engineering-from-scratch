Predict both results and their shapes.

```python
a = Matrix([[1, 2], [3, 4]])
b = Matrix([[5, 6], [7, 8]])
(a.element_wise_multiply(b).data, (a @ b).data)
```

## Dense-layer shape trace

`(4,3) @ (3,1) + (4,1)` produces `(4,1)`. A row or column bias is explicit broadcasting; a mismatched product must raise `ValueError`.

```python
x = Matrix([[0.5], [0.8], [0.2]])
w = Matrix([[1, 0, 0], [0, 1, 0], [0.5, 0.5, 0], [0, 0, 1]])
relu_matrix(w @ x).shape
```
