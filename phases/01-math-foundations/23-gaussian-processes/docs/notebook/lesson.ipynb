# Gaussian Processes from Scratch

This notebook follows the NumPy implementation in `code/main.py` and keeps the training/test fixtures local.

```python
from pathlib import Path
import sys
candidates = (Path.cwd() / 'code', Path.cwd().parents[1] / 'code', Path('phases/01-math-foundations/23-gaussian-processes/code'))
for candidate in candidates:
    if (candidate / 'main.py').exists():
        sys.path.insert(0, str(candidate))
        break
import numpy as np
import main
x_train = np.array([-2.0, -1.3, -0.4, 0.2, 1.0, 1.8])
y_train = np.sin(x_train) + np.array([0.02, -0.04, 0.03, -0.01, 0.04, -0.02])
```

```python
kernel = main.rbf_kernel(x_train, x_train, length_scale=0.6)
assert kernel.shape == (6, 6)
np.testing.assert_allclose(kernel, kernel.T)
print('diagonal:', np.diag(kernel))
mean, variance, jitter = main.gp_posterior(x_train, y_train, [-3.0, 0.2, 3.0], length_scale=0.6, noise_variance=0.01)
print('mean:', mean)
print('variance:', variance, 'jitter:', jitter)
```

```python
chosen, scores = main.select_length_scale(x_train, y_train, [0.25, 0.6, 1.2, 2.0])
assert chosen in scores
print('candidate scores:', scores)
print('selected length scale:', chosen)
print('near/far variance:', variance[1], variance[0])
```
