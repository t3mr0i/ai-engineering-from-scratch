## Build It: vectors and projections

Compute the dot product, norm, and projection before looking at a downstream application.

```python
a = Vector([3, 4])
b = Vector([1, 0])
projection = a.project_onto(b)
(a.magnitude(), projection.components, (a - projection).dot(b))
```

The expected tuple is `(5.0, [3.0, 0.0], 0.0)`: the residual is orthogonal to the target direction.

```python
basis = gram_schmidt([Vector([1, 1]), Vector([1, 0])])
[(round(v.magnitude(), 6), round(basis[0].dot(v), 6)) for v in basis]
Matrix([[1, 0, 0], [0, 1, 0]]) @ Vector([2, -1, 7])
```
