# Norms and Distances

The notebook uses the stdlib metrics in `code/distances.py` to make a retrieval choice explicit.

```python
from pathlib import Path
import sys
candidates = (Path.cwd() / 'code', Path.cwd().parents[1] / 'code', Path.cwd() / 'phases/01-math-foundations/14-norms-and-distances/code')
code_dir = next(path for path in candidates if (path / 'distances.py').is_file())
sys.path.insert(0, str(code_dir))
from distances import l1_norm, l2_norm, cosine_similarity, cosine_distance, jaccard_similarity, edit_distance, wasserstein_1d, find_k_nearest
```

## Build It: vector metrics

For `[3,-4]`, L1 is 7 and L2 is 5. Cosine compares direction, so `[1,0]` and `[2,0]` have similarity 1 and distance 0.

```python
l1_norm([3, -4]), l2_norm([3, -4]), cosine_similarity([1, 0], [2, 0]), cosine_distance([1, 0], [2, 0])
```

## Use It: discrete evidence and distributions

Jaccard uses intersection over union for sets. Edit distance counts insertions, deletions, and substitutions. The one-dimensional Wasserstein routine compares cumulative mass across equal bins.

```python
jaccard_similarity({'a', 'b', 'c'}, {'b', 'c', 'd'}), edit_distance('kitten', 'sitting'), wasserstein_1d([1.0, 0.0], [0.0, 1.0])
```

## Ship It: nearest-neighbor contract

Pass a metric explicitly to `find_k_nearest`; the result is sorted `(index, distance)` pairs, not labels or embeddings.

```python
points = [[0, 0], [3, 4], [1, 1]]
squared_l2 = lambda a, b: sum((x - y) ** 2 for x, y in zip(a, b))
find_k_nearest([0, 0], points, squared_l2, k=2)
```
