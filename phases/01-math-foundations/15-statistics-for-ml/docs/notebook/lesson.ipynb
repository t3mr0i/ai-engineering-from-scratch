## Build It: describe a fixture

The implementation uses sample variance by default, so `[1,2,3,4]` has variance `5/3`. Percentiles are linearly interpolated between sorted observations.

```python
data = [1, 2, 3, 4]
mean(data), median(data), variance(data), percentile(data, 75)
```

## Use It: association and hypothesis tests

Pearson detects linear association; Spearman applies Pearson to ranks. A p-value is a tail probability conditional on the null, not the probability that the null is true.

```python
x = [1, 2, 3, 4]
pearson_correlation(x, [2, 4, 6, 8]), spearman_correlation(x, [1, 4, 9, 16]), one_sample_ttest([10, 11, 12, 13, 14], mu_0=0)
```

## Ship It: uncertainty plus effect size

Bootstrap resamples the observed list with replacement. Pair its interval with Cohen's d: statistical significance and practical magnitude answer different questions.

```python
import random
random.seed(3)
interval = bootstrap_statistic([1, 2, 3, 4, 5], mean, n_bootstrap=250, ci=90)
interval, cohens_d([0, 1, 2], [2, 3, 4])
```
